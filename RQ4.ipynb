```python
# RQ4: Custom CNN vs MobileNetV2 Transfer Learning

import os
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization, GlobalAveragePooling2D
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.applications import MobileNetV2


# 1. Create output folders

OUTPUT_DIR = Path("Results_RQ_Outputs")
BASE_DIR = OUTPUT_DIR / "RQ4_CNN_vs_MobileNetV2"
FIG_DIR = BASE_DIR / "figures"
TABLE_DIR = BASE_DIR / "tables"
TABLE_PNG_DIR = BASE_DIR / "tables_png"
MODEL_DIR = BASE_DIR / "models"

for folder in [OUTPUT_DIR, BASE_DIR, FIG_DIR, TABLE_DIR, TABLE_PNG_DIR, MODEL_DIR]:
    folder.mkdir(parents=True, exist_ok=True)


# 2. Load CSV files

styles = pd.read_csv("styles.csv", on_bad_lines="skip")
images = pd.read_csv("images.csv")

styles["id"] = styles["id"].astype(str)
styles["filename"] = styles["id"] + ".jpg"

if "filename" in images.columns:
    df = pd.merge(styles, images, on="filename", how="inner")
else:
    df = styles.copy()


# 3. Prepare image paths

IMAGE_FOLDER = Path("images")

df["image_path"] = df["filename"].apply(lambda x: str(IMAGE_FOLDER / x))
df = df[df["image_path"].apply(os.path.exists)].copy()

if len(df) == 0:
    raise ValueError("No images found. Check that JPG files are inside the images folder.")


# 4. Select top 5 article categories

df = df.dropna(subset=["articleType"])

top_classes = df["articleType"].value_counts().head(5).index
df = df[df["articleType"].isin(top_classes)].copy()

print("Selected classes:")
print(df["articleType"].value_counts())


# 5. Dataset description table

dataset_table = pd.DataFrame({
    "Description": [
        "Research Question",
        "Total images used",
        "Target column",
        "Selected classes",
        "Input image size",
        "Models compared"
    ],
    "Value": [
        "RQ4: How does a custom CNN compare with MobileNetV2 transfer learning?",
        len(df),
        "articleType",
        ", ".join(top_classes),
        "128 x 128",
        "Custom CNN and MobileNetV2"
    ]
})

dataset_table.to_csv(TABLE_DIR / "rq4_dataset_description.csv", index=False)

plt.figure(figsize=(13, 4))
plt.axis("off")
plt.table(
    cellText=dataset_table.values,
    colLabels=dataset_table.columns,
    cellLoc="center",
    loc="center"
)
plt.title("RQ4 Dataset Description Table")
plt.tight_layout()
plt.savefig(TABLE_PNG_DIR / "rq4_dataset_description.png", dpi=300)
plt.show()


# 6. Class distribution

class_counts = df["articleType"].value_counts()

class_distribution = class_counts.reset_index()
class_distribution.columns = ["Article Type", "Number of Images"]
class_distribution.to_csv(TABLE_DIR / "rq4_class_distribution.csv", index=False)

plt.figure(figsize=(8, 5))
class_counts.plot(kind="bar")
plt.title("RQ4 Class Distribution")
plt.xlabel("Article Type")
plt.ylabel("Number of Images")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(FIG_DIR / "rq4_class_distribution.png", dpi=300)
plt.show()

plt.figure(figsize=(8, 3))
plt.axis("off")
plt.table(
    cellText=class_distribution.values,
    colLabels=class_distribution.columns,
    cellLoc="center",
    loc="center"
)
plt.title("RQ4 Class Distribution Table")
plt.tight_layout()
plt.savefig(TABLE_PNG_DIR / "rq4_class_distribution_table.png", dpi=300)
plt.show()


# 7. Train validation test split

train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    stratify=df["articleType"],
    random_state=42
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["articleType"],
    random_state=42
)

split_table = pd.DataFrame({
    "Split": ["Training", "Validation", "Testing"],
    "Number of Images": [len(train_df), len(val_df), len(test_df)],
    "Percentage": [
        round(len(train_df) / len(df) * 100, 2),
        round(len(val_df) / len(df) * 100, 2),
        round(len(test_df) / len(df) * 100, 2)
    ]
})

split_table.to_csv(TABLE_DIR / "rq4_data_split.csv", index=False)

plt.figure(figsize=(7, 3))
plt.axis("off")
plt.table(
    cellText=split_table.values,
    colLabels=split_table.columns,
    cellLoc="center",
    loc="center"
)
plt.title("RQ4 Data Split Table")
plt.tight_layout()
plt.savefig(TABLE_PNG_DIR / "rq4_data_split.png", dpi=300)
plt.show()


# 8. Image generators

IMG_SIZE = (128, 128)
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.2,
    horizontal_flip=True
)

test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_dataframe(
    train_df,
    x_col="image_path",
    y_col="articleType",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=True
)

val_generator = test_datagen.flow_from_dataframe(
    val_df,
    x_col="image_path",
    y_col="articleType",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

test_generator = test_datagen.flow_from_dataframe(
    test_df,
    x_col="image_path",
    y_col="articleType",
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)

class_names = list(train_generator.class_indices.keys())
num_classes = len(class_names)


# 9. Build custom CNN model

custom_cnn = Sequential([
    Conv2D(32, (3, 3), activation="relu", input_shape=(128, 128, 3)),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    Conv2D(64, (3, 3), activation="relu"),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    Conv2D(128, (3, 3), activation="relu"),
    BatchNormalization(),
    MaxPooling2D(2, 2),

    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.5),
    Dense(num_classes, activation="softmax")
])

custom_cnn.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

custom_cnn.summary()


# 10. Build MobileNetV2 transfer learning model

mobilenet_base = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(128, 128, 3)
)

mobilenet_base.trainable = False

x = mobilenet_base.output
x = GlobalAveragePooling2D()(x)
x = Dense(128, activation="relu")(x)
x = Dropout(0.5)(x)
output = Dense(num_classes, activation="softmax")(x)

mobilenet_model = Model(inputs=mobilenet_base.input, outputs=output)

mobilenet_model.compile(
    optimizer=Adam(learning_rate=0.0001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

mobilenet_model.summary()


# 11. Save architecture comparison table

architecture_table = pd.DataFrame({
    "Model": [
        "Custom CNN",
        "MobileNetV2 Transfer Learning"
    ],
    "Architecture": [
        "Three Conv2D blocks followed by Flatten, Dense, Dropout, Softmax",
        "Pre-trained MobileNetV2 base with GlobalAveragePooling, Dense, Dropout, Softmax"
    ],
    "Pre-trained Weights": [
        "No",
        "Yes, ImageNet"
    ],
    "Trainable Feature Extractor": [
        "Yes",
        "No, frozen MobileNetV2 base"
    ],
    "Learning Rate": [
        "0.001",
        "0.0001"
    ]
})

architecture_table.to_csv(TABLE_DIR / "rq4_model_architecture_comparison.csv", index=False)

plt.figure(figsize=(14, 3))
plt.axis("off")
plt.table(
    cellText=architecture_table.values,
    colLabels=architecture_table.columns,
    cellLoc="center",
    loc="center"
)
plt.title("RQ4 Model Architecture Comparison")
plt.tight_layout()
plt.savefig(TABLE_PNG_DIR / "rq4_model_architecture_comparison.png", dpi=300)
plt.show()


# 12. Train custom CNN

history_custom = custom_cnn.fit(
    train_generator,
    validation_data=val_generator,
    epochs=2
)

custom_cnn.save(MODEL_DIR / "rq4_custom_cnn_model.h5")


# 13. Train MobileNetV2

history_mobilenet = mobilenet_model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=10
)

mobilenet_model.save(MODEL_DIR / "rq4_mobilenetv2_model.h5")


# 14. Evaluate custom CNN

test_generator.reset()

custom_pred_probs = custom_cnn.predict(test_generator)
custom_pred_classes = np.argmax(custom_pred_probs, axis=1)
true_classes = test_generator.classes

custom_test_accuracy = accuracy_score(true_classes, custom_pred_classes)

print("Custom CNN Test Accuracy:", custom_test_accuracy)


# 15. Evaluate MobileNetV2

test_generator.reset()

mobilenet_pred_probs = mobilenet_model.predict(test_generator)
mobilenet_pred_classes = np.argmax(mobilenet_pred_probs, axis=1)

mobilenet_test_accuracy = accuracy_score(true_classes, mobilenet_pred_classes)

print("MobileNetV2 Test Accuracy:", mobilenet_test_accuracy)


# 16. Training accuracy comparison curve

plt.figure(figsize=(8, 5))
plt.plot(history_custom.history["accuracy"], label="Custom CNN Training Accuracy")
plt.plot(history_mobilenet.history["accuracy"], label="MobileNetV2 Training Accuracy")
plt.title("RQ4 Training Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "rq4_training_accuracy_comparison.png", dpi=300)
plt.show()


# 17. Validation accuracy comparison curve

plt.figure(figsize=(8, 5))
plt.plot(history_custom.history["val_accuracy"], label="Custom CNN Validation Accuracy")
plt.plot(history_mobilenet.history["val_accuracy"], label="MobileNetV2 Validation Accuracy")
plt.title("RQ4 Validation Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "rq4_validation_accuracy_comparison.png", dpi=300)
plt.show()


# 18. Training loss comparison curve

plt.figure(figsize=(8, 5))
plt.plot(history_custom.history["loss"], label="Custom CNN Training Loss")
plt.plot(history_mobilenet.history["loss"], label="MobileNetV2 Training Loss")
plt.title("RQ4 Training Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "rq4_training_loss_comparison.png", dpi=300)
plt.show()


# 19. Validation loss comparison curve

plt.figure(figsize=(8, 5))
plt.plot(history_custom.history["val_loss"], label="Custom CNN Validation Loss")
plt.plot(history_mobilenet.history["val_loss"], label="MobileNetV2 Validation Loss")
plt.title("RQ4 Validation Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.tight_layout()
plt.savefig(FIG_DIR / "rq4_validation_loss_comparison.png", dpi=300)
plt.show()


# 20. Test accuracy bar chart

accuracy_labels = ["Custom CNN", "MobileNetV2"]
accuracy_values = [custom_test_accuracy, mobilenet_test_accuracy]

plt.figure(figsize=(7, 5))
plt.bar(accuracy_labels, accuracy_values)
plt.title("RQ4 Test Accuracy Comparison")
plt.xlabel("Model")
plt.ylabel("Test Accuracy")
plt.ylim(0, 1)

for i, value in enumerate(accuracy_values):
    plt.text(i, value + 0.01, str(round(value * 100, 2)) + "%", ha="center")

plt.tight_layout()
plt.savefig(FIG_DIR / "rq4_test_accuracy_comparison.png", dpi=300)
plt.show()


# 21. Classification reports

custom_report = classification_report(
    true_classes,
    custom_pred_classes,
    target_names=class_names,
    output_dict=True,
    zero_division=0
)

mobilenet_report = classification_report(
    true_classes,
    mobilenet_pred_classes,
    target_names=class_names,
    output_dict=True,
    zero_division=0
)

custom_report_df = pd.DataFrame(custom_report).transpose()
mobilenet_report_df = pd.DataFrame(mobilenet_report).transpose()

custom_report_df.to_csv(TABLE_DIR / "rq4_classification_report_custom_cnn.csv")
mobilenet_report_df.to_csv(TABLE_DIR / "rq4_classification_report_mobilenetv2.csv")

plt.figure(figsize=(12, 6))
plt.axis("off")
plt.table(
    cellText=np.round(custom_report_df.values, 3),
    rowLabels=custom_report_df.index,
    colLabels=custom_report_df.columns,
    cellLoc="center",
    loc="center"
)
plt.title("RQ4 Classification Report - Custom CNN")
plt.tight_layout()
plt.savefig(TABLE_PNG_DIR / "rq4_classification_report_custom_cnn.png", dpi=300)
plt.show()

plt.figure(figsize=(12, 6))
plt.axis("off")
plt.table(
    cellText=np.round(mobilenet_report_df.values, 3),
    rowLabels=mobilenet_report_df.index,
    colLabels=mobilenet_report_df.columns,
    cellLoc="center",
    loc="center"
)
plt.title("RQ4 Classification Report - MobileNetV2")
plt.tight_layout()
plt.savefig(TABLE_PNG_DIR / "rq4_classification_report_mobilenetv2.png", dpi=300)
plt.show()


# 22. Confusion matrices

custom_cm = confusion_matrix(true_classes, custom_pred_classes)
mobilenet_cm = confusion_matrix(true_classes, mobilenet_pred_classes)

custom_cm_df = pd.DataFrame(custom_cm, index=class_names, columns=class_names)
mobilenet_cm_df = pd.DataFrame(mobilenet_cm, index=class_names, columns=class_names)

custom_cm_df.to_csv(TABLE_DIR / "rq4_confusion_matrix_custom_cnn.csv")
mobilenet_cm_df.to_csv(TABLE_DIR / "rq4_confusion_matrix_mobilenetv2.csv")

plt.figure(figsize=(8, 6))
plt.imshow(custom_cm)
plt.title("RQ4 Confusion Matrix - Custom CNN")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.xticks(range(len(class_names)), class_names, rotation=45)
plt.yticks(range(len(class_names)), class_names)

for i in range(len(class_names)):
    for j in range(len(class_names)):
        plt.text(j, i, custom_cm[i, j], ha="center", va="center")

plt.colorbar()
plt.tight_layout()
plt.savefig(FIG_DIR / "rq4_confusion_matrix_custom_cnn.png", dpi=300)
plt.show()

plt.figure(figsize=(8, 6))
plt.imshow(mobilenet_cm)
plt.title("RQ4 Confusion Matrix - MobileNetV2")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.xticks(range(len(class_names)), class_names, rotation=45)
plt.yticks(range(len(class_names)), class_names)

for i in range(len(class_names)):
    for j in range(len(class_names)):
        plt.text(j, i, mobilenet_cm[i, j], ha="center", va="center")

plt.colorbar()
plt.tight_layout()
plt.savefig(FIG_DIR / "rq4_confusion_matrix_mobilenetv2.png", dpi=300)
plt.show()


# 23. Final comparison table

custom_final_train_acc = history_custom.history["accuracy"][-1]
custom_final_val_acc = history_custom.history["val_accuracy"][-1]

mobilenet_final_train_acc = history_mobilenet.history["accuracy"][-1]
mobilenet_final_val_acc = history_mobilenet.history["val_accuracy"][-1]

comparison_table = pd.DataFrame({
    "Model": ["Custom CNN", "MobileNetV2 Transfer Learning"],
    "Final Training Accuracy": [
        round(custom_final_train_acc, 4),
        round(mobilenet_final_train_acc, 4)
    ],
    "Final Validation Accuracy": [
        round(custom_final_val_acc, 4),
        round(mobilenet_final_val_acc, 4)
    ],
    "Test Accuracy": [
        round(custom_test_accuracy, 4),
        round(mobilenet_test_accuracy, 4)
    ],
    "Generalization Gap": [
        round(custom_final_train_acc - custom_final_val_acc, 4),
        round(mobilenet_final_train_acc - mobilenet_final_val_acc, 4)
    ],
    "Accuracy Percentage": [
        str(round(custom_test_accuracy * 100, 2)) + "%",
        str(round(mobilenet_test_accuracy * 100, 2)) + "%"
    ]
})

comparison_table.to_csv(TABLE_DIR / "rq4_final_model_comparison.csv", index=False)

plt.figure(figsize=(14, 3))
plt.axis("off")
plt.table(
    cellText=comparison_table.values,
    colLabels=comparison_table.columns,
    cellLoc="center",
    loc="center"
)
plt.title("RQ4 Final Model Performance Comparison")
plt.tight_layout()
plt.savefig(TABLE_PNG_DIR / "rq4_final_model_comparison.png", dpi=300)
plt.show()


# 24. Final interpretation table

if mobilenet_test_accuracy > custom_test_accuracy:
    finding = "MobileNetV2 performed better than the custom CNN because it uses pre-trained ImageNet features."
elif mobilenet_test_accuracy < custom_test_accuracy:
    finding = "The custom CNN performed better than MobileNetV2 for this selected dataset."
else:
    finding = "Both models achieved similar test accuracy."

interpretation_table = pd.DataFrame({
    "Research Question": [
        "How does the performance of a custom CNN compare with MobileNetV2 transfer learning?"
    ],
    "Main Finding": [finding],
    "Custom CNN Accuracy": [str(round(custom_test_accuracy * 100, 2)) + "%"],
    "MobileNetV2 Accuracy": [str(round(mobilenet_test_accuracy * 100, 2)) + "%"],
    "Interpretation": [
        "Transfer learning usually improves performance because MobileNetV2 has already learned useful visual features from a large image dataset."
    ]
})

interpretation_table.to_csv(TABLE_DIR / "rq4_final_interpretation.csv", index=False)

plt.figure(figsize=(15, 3))
plt.axis("off")
plt.table(
    cellText=interpretation_table.values,
    colLabels=interpretation_table.columns,
    cellLoc="center",
    loc="center"
)
plt.title("RQ4 Final Interpretation")
plt.tight_layout()
plt.savefig(TABLE_PNG_DIR / "rq4_final_interpretation.png", dpi=300)
plt.show()


# 25. Save training history

history_table = pd.DataFrame({
    "Epoch": list(range(1, len(history_custom.history["accuracy"]) + 1)),
    "Custom CNN Training Accuracy": history_custom.history["accuracy"],
    "Custom CNN Validation Accuracy": history_custom.history["val_accuracy"],
    "Custom CNN Training Loss": history_custom.history["loss"],
    "Custom CNN Validation Loss": history_custom.history["val_loss"],
    "MobileNetV2 Training Accuracy": history_mobilenet.history["accuracy"],
    "MobileNetV2 Validation Accuracy": history_mobilenet.history["val_accuracy"],
    "MobileNetV2 Training Loss": history_mobilenet.history["loss"],
    "MobileNetV2 Validation Loss": history_mobilenet.history["val_loss"]
})

history_table.to_csv(TABLE_DIR / "rq4_training_history.csv", index=False)

print("RQ4 completed successfully.")
print("All outputs saved in:", BASE_DIR)
print("Figures saved in:", FIG_DIR)
print("CSV tables saved in:", TABLE_DIR)
print("PNG tables saved in:", TABLE_PNG_DIR)
print("Models saved in:", MODEL_DIR)
```
